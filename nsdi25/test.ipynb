```python
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerLine2D

# 绘制数据
lines = [plt.plot([1, 2, 3], [i, i+1, i+2], label=f'Line {i+1}')[0] for i in range(6)]

# 创建图例
legend = plt.legend(lines[0:3], [f'Line {i+1}' for i in range(3)], loc='upper left', ncol = 3)
plt.gca().add_artist(legend)

legend = plt.legend(lines[3:6], [f'Line {i+1}' for i in range(3, 6)], loc='upper left', ncol = 3, bbox_to_anchor=(0, -0.15))

plt.show()
```
